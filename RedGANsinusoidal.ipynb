```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
# Una señal sinusoidal presenta la siguiente fórmula: y(t)=amp*sin(2*π*frec*t+ϕ)

# amp: amplitud
# frec: frecuencia
# ϕ: fase
# t: tiempo
```

```python
# Para crear una señal de forma sinusoidal en primer lugar es necesario fijar una serie de parámetros que definan la forma de esta señal

amp = 1.0  # Valor máximo que alcanza la señal seno, hace que la señal se prolongue más o menos en el eje Y. Puede ser un valor entero o decimal y se puede medir en la unidad que queramos
frec = 5  # Número de ciclos (oscilaciones completas) que tienen lugar por segundo Se mide en hercios (Hz) (+ frecuencia, + comprimida la señal en el eje del tiempo, - amplitud)
fase = 0        # Desplazamiento horizontal de la señal, punto del eje X en que empieza la onda. Se mide en radianes (rad)
dur = 1.0  # Tiempo que dura la señal seno. Se mide en segundos (s). Puede ser un valor entero o decimal
frec_muest = 1000       #Frecuencia de muestreo, número de veces que se mide el valor de la señal por segundo. Se mide en hercios (Hz)


# La frecuencia de muestreo (frec_muest) tiene que ser al menos el doble de la frecuencia de la propia señal, siguiendo el teorema de Nyquist, para evitar el aliasing y la pérdida de información

# La frecuencia y la fase también pueden ser valores decimales, aunque por la naturaleza de los propios parámetros no es tan común
```

```python
# Para poder representar la función de la ecuación sinusoidal (y(t)) es necesario crear un vector tiempo, es decir, los puntos del eje X en los que vamos a representar valores

# Vamos a tener un total de puntos igual a la frecuencia de muestreo (muestras por segundo) por la duración total de nuestra señal seno
puntos = frec_muest * dur

# Creamos un vector (array 1D) con un total de puntos desde el inicio (0) hasta el final (dur) con un total de tantos números en ese rango (puntos) sin incluir el valor final
t = np.linspace (0, dur, int(puntos), endpoint=False)
```

```python
# Creamos la señal seno siguiendo la ecuación que hemos indicado en un primer momento

señal_y = amp * np.sin (2 * np.pi * frec * t + fase)
```

```python
# Representamos por pantalla la señal seno que hemos creado, aunque podemos modificar los parámetros de la misma según nuestras preferencias

plt.plot (t, señal_y)
plt.title ("Señal Senosoidal (y(t))")
plt.xlabel ("Tiempo de la señal (segundos)")
plt.ylabel ("Amplitud de la señal")
plt.grid (False)
plt.show()
```

```python
# Podemos representar la señal sobre una cuadrícula para realizar las medidas y consideraciones necesarias sobre la señal

plt.plot (t, señal_y)
plt.title ("Señal Senosoidal (y(t))")
plt.xlabel ("Tiempo de la señal (segundos)")
plt.ylabel ("Amplitud de la señal")
plt.grid (True)
plt.show()
```
